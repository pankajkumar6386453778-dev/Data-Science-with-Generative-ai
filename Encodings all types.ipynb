# Target-Guided Ordinal Encoding (Definition):

Target-Guided Ordinal Encoding is a technique used in machine learning to convert categorical variables into numerical values by using the target variable (output) to determine the order of categories.

🔑 Simple Definition:

It assigns a rank (order) to each category based on the mean (or another statistic) of the target variable for that category.

```python
import pandas as pd

#create a sample dataframe with categorical variable and target variable
df = pd.DataFrame({
    'city':['New York' ,'London','Paris','Tokyo','New York','Paris'],
    'price':[200,150,300,250,180,320]
})
```

```python
df.head()
```

```python
## calculate mean price of each city
mean_price =df.groupby('city')['price'].mean().to_dict()
mean_price
```

```python
## replace each city with its mean price
df['city_encoded']=df['city'].map(mean_price)
```

## Nominal /One Hot Encoding Techniques

#Nominal One-Hot Encoding is a technique used in machine learning to convert nominal categorical data (categories with no natural order) into numerical format by creating separate binary columns (0 or 1) for each category.

# disadvantages
1.create sparse matrix[overfitting]
2.it will increase the no of features

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
#cretae a sample dataframe with categorical variable
df = pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
df
```

```python
## create an instance of one hot encoder
encoder = OneHotEncoder()
```

```python
#fit the encoder to the dataframe and transform the categorical variable
encoded = encoder.fit_transform(df[["color"]])
```

```python
encoded
```

```python
import pandas as pd
encoded_df = pd.DataFrame(encoded.toarray(),columns=encoder.get_feature_names_out())
```

```python
encoder.get_feature_names_out()
```

```python
encoded_df
```

```python
pd.concat([df,encoded_df],axis=1)
```

#Label Encoding (Ordinal Encoder) – Definition

Label Encoding (also called Ordinal Encoding) is a technique used in machine learning to convert categorical data (text labels) into numerical values by assigning each unique category a number.

🔹 Simple Definition:

Label/Ordinal Encoding transforms categories into numbers like:

Category → Number

## Label Encoder

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
#cretae a sample dataframe with categorical variable
df = pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
df
```

```python
## create instance of LabelEncoder
encoder = LabelEncoder()
```

```python
encoder.fit_transform(df['color'])
```

## Ordinal Encoding

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

#create a sample dataframe with an ordinal variable
df = pd.DataFrame({
    'size':['small','medium','large','medium','small','large']
})
```

```python
df
```

```python
# create an instance of the ordinalencoder class and fit transform
encoder = OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df[['size']])
```
