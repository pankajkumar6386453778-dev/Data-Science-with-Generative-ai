## Chi Square TestWith Python

```python
import scipy.stats as stat
import numpy as np
```

```python
## number of hours student study in weekly basis daily
expected_data=[8,6,7,9,6,9,7]
observed_data=[7,8,6,9,9,6,7]
```

```python
sum(expected_data),sum(observed_data)
```

```python
## chi square gooness og fit test
chisquare_test_statistics,p_value=stat.chisquare(observed_data ,expected_data)
```

```python
chisquare_test_statistics,p_value
```

```python
## find the critical value
significance_value=0.05
dof =len(expected_data)-1
print(dof)
critical_value =stat.chi2.ppf(1-significance_value,dof)
```

```python
critical_value
```

```python
if chisquare_test_statistics>critical_value:
    print("Rejected the null hypothesis")
else:
    print("Accept the null  Hypothesis")
```
