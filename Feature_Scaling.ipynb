```python
total_bill = list(df["total_bill"])
```

```python
import numpy as np
mean = np.mean(total_bill)
std = np.std(total_bill)
```

```python
mean , std
```

```python
normalized_data = []
for i in total_bill:
    z_score=(i-mean)/std # z-score
    normalized_data.append(z_score)

print(normalized_data)
```

```python
min_max=MinMaxScaler()
```
