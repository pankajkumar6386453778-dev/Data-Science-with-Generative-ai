## F test with python

```python
#Evidence to say that two population variance are not equal

worker1 = [18,19,22,25,27,28,41,45,51,55]
worker2=[14,15,15,17,18,22,25,25,27,34]
```

```python
##calculatin f test
import numpy as np
f_test= np.var(worker1)/np.var(worker2)
```

```python
f_test
```

```python
## Degree of Freedom
df1 =len(worker1)-1
df2=len(worker2)-1
significance_value =0.05
```

```python
import scipy.stats as stat
```

```python
critical_values = stat.f.ppf(q=1-significance_value, dfn=df1,dfd=df2)
```

```python
critical_values
```

```python
if f_test>critical_values:
    print("Reject the Null Hypothesis")
else:
    print("We faill to reject the Null Hypothesis")
```
